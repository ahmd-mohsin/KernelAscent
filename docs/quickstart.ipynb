# KernelAscent quickstart
Install, then evaluate a model on Tier-1/2. Generation is Bedrock API (no GPU); grading needs a GPU.

```python
!pip install "kernelascent @ git+https://github.com/ahmd-mohsin/KernelAscent"
```

```python
import os
os.environ['AWS_PROFILE'] = 'bedrock'  # a profile with Bedrock access (us-east-1)
# API generation only (no GPU): candidate kernels + reasoning trajectories
!kernelascent gen --model us.anthropic.claude-opus-4-8 --tiers L1,L2 --k 1 --out runs/opus
```

```python
# Grade on a GPU runtime (Runtime > Change runtime type > GPU), then:
!pip install torch triton
!kernelascent grade --candir runs/opus --out runs/opus/summary.json
import json; print(json.load(open('runs/opus/summary.json'))['tasks'][:3])
```
